###Feature Scaling

```python
x=df.drop(columns="class")
y=df["class"]
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
df_x=sc.fit_transform(x)
```

```python
df_x
```

###Splitting the dataset

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(x,y,test_size=0.2,random_state=1)
```

### Fixing Data Leakage in Feature Scaling

Currently, `StandardScaler` is applied to the entire dataset (`x`) before the `train_test_split`. This leads to **data leakage**, where information from the test set influences the scaling of the training set. This can artificially boost model performance during evaluation but leads to poor generalization on new, unseen data.

To correct this, we need to apply `StandardScaler` *after* the `train_test_split`. The scaler should be `fit` only on the training data (`xtrain`) and then `transform` both the training and testing data (`xtrain` and `xtest`).

```python
# Re-apply StandardScaler to prevent data leakage
sc_new = StandardScaler()
xtrain_scaled = sc_new.fit_transform(xtrain)
xtest_scaled = sc_new.transform(xtest)

print("Shape of scaled training features:", xtrain_scaled.shape)
print("Shape of scaled testing features:", xtest_scaled.shape)
```

###label encoding

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
ytrain=le.fit_transform(ytrain)
ytest=le.transform(ytest)
```

####using svm(suppprt vector machine)

```python
from sklearn.svm import SVC
```

```python
model=SVC(kernel="rbf")
```

```python
model.fit(xtrain_scaled,ytrain)
```

```python
y_predict=model.predict(xtest_scaled)
```

```python
y_predict
```

##checking the accuracy score

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy=accuracy_score(ytest,y_predict)
```

```python
accuracy
```

```python
cm=confusion_matrix(ytest,y_predict)
sns.heatmap(cm,annot=True)
plt.xlabel("predicted")
plt.ylabel("actual")
plt.title("confusion matrix")
```

```python
print(classification_report(ytest,y_predict))
```
